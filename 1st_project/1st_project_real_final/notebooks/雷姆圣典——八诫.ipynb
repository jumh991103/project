## 雷姆圣典-八诫
1. 成心相信并只视雷姆为唯一至高真神。
진심으로 렘을 믿으며, 오직 렘만을 유일하고 지고한 참된 신으로 여긴다.
2. 包容并接受任何性别，种族，国家等，甚至是信仰过期他宗教派别的认识加入。
성별, 인종, 국가 등에 관계없이 누구든 포용하고 받아들이며, 심지어 과거 다른 종교나 종파를 믿었던 사람도 받아들인다.
3. 每位教众都有传教，宋扬雷姆教的权利。
모든 신도는 렘교를 전파하고 찬양할 권리가 있다.
4. 在传教的同时，必须理性地说出语等的教义，信念，若传教对象尚未理解或是不接受福音的恩赐，不得强迫他人。
포교할 때에는 우리의 교리와 신념을 이성적으로 설명해야 하며, 상대방이 아직 이해하지 못했거나 복음의 은혜를 받아들이지 않는다면 타인에게 강요해서는 안 된다.
5. 不得攻击、污蔑其他宗教派别，甚至是口出恶言，而是贯彻雷姆真神的包容，以示雷姆教的宽阔胸襟。
다른 종교나 종파를 공격하거나 비방해서는 안 되며, 악언을 해서도 안 된다. 오히려 렘 참신의 포용을 실천하여 렘교의 넓은 아량을 보여야 한다.
6. 了解并思考雷姆真神在人间的所作所为所说，由教众们自行挑选其中对自身有影响力的，并激励、反思在现实生活中的自己。
렘 참신이 인간 세상에서 행하고 말한 것들을 이해하고 생각하며, 신도들은 그중 자신에게 영향을 주는 것을 스스로 선택하여 현실 속의 자신을 격려하고 성찰한다.
7. 即使在那位黑发少年的心中，雷姆真神可能不是唯一，要知道，祂依然永存于我们心中。
비록 그 검은 머리 소년의 마음속에서 렘 참신이 유일한 존재가 아닐지라도, 렘은 여전히 우리의 마음속에 영원히 존재한다는 것을 알아야 한다.
8. 温柔二字，便是雷姆的化身。
「다정함/온유함」이라는 두 글자야말로 렘의 화신이다.
